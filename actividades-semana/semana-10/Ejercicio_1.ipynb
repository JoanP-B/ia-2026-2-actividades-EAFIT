```python
import numpy as np
import matplotlib.pyplot as plt

# Semilla para obtener siempre los mismos resultados
np.random.seed(42)

# Parámetros de la recta y = mx + b
m = 2.5
b = 4

# Generar 100 coordenadas x aleatorias
x = np.random.uniform(-10, 10, 100)

# Agregar ruido normal para dispersar los puntos alrededor de la recta
ruido = np.random.normal(loc=0, scale=3, size=100)
y = m * x + b + ruido

# Graficar los puntos generados y la recta ideal
plt.figure(figsize=(10, 6))
plt.scatter(x, y, color="royalblue", alpha=0.75, label="Muestras con ruido")

x_recta = np.linspace(-10, 10, 200)
y_recta = m * x_recta + b
plt.plot(x_recta, y_recta, color="crimson", linewidth=2, label=f"Recta ideal: y = {m}x + {b}")

plt.title("100 muestras aleatorias alrededor de una recta")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(alpha=0.3)
plt.legend()
plt.show()

print(f"Cantidad de muestras generadas: {len(x)}")
```
