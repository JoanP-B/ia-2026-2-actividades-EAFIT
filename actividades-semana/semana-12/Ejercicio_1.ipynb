```python
%pip install -q sentence-transformers scikit-learn matplotlib

from sentence_transformers import SentenceTransformer
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Modelo multilingue; la primera ejecucion puede descargarlo de Internet.
modelo = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")

def convertir_a_vector(texto):
    """Convierte un texto en un vector numerico (embedding)."""
    if not isinstance(texto, str) or not texto.strip():
        raise ValueError("El texto debe ser una cadena no vacia.")

    return modelo.encode(
        texto.strip(),
        convert_to_numpy=True,
        normalize_embeddings=True
    )

def graficar_embeddings(textos, vectores):
    """Proyecta embeddings a 2 dimensiones para visualizar su relacion."""
    if len(textos) < 2:
        raise ValueError("Se necesitan al menos dos textos para graficar.")
    if len(vectores) != len(textos):
        raise ValueError("Debe haber un vector por cada texto.")

    coordenadas = PCA(n_components=2).fit_transform(vectores)

    plt.figure(figsize=(10, 6))
    plt.scatter(coordenadas[:, 0], coordenadas[:, 1], s=90, color="teal")

    for indice, texto_etiqueta in enumerate(textos):
        etiqueta = texto_etiqueta if len(texto_etiqueta) <= 40 else texto_etiqueta[:37] + "..."
        plt.annotate(
            etiqueta,
            (coordenadas[indice, 0], coordenadas[indice, 1]),
            xytext=(6, 6),
            textcoords="offset points"
        )

    plt.title("Espacio de embeddings de texto (proyeccion PCA)")
    plt.xlabel("Componente principal 1")
    plt.ylabel("Componente principal 2")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# Ingresa un texto para convertirlo en embedding.
texto = input("Escribe un texto: ").strip()
if not texto:
    raise ValueError("El texto no puede estar vacio.")

# Se compara el texto ingresado con frases de temas distintos.
textos = [
    texto,
    "La inteligencia artificial permite analizar informacion.",
    "Me gusta preparar comida italiana.",
    "El aprendizaje automatico encuentra patrones en los datos."
 ]

# Codificar todo en una sola llamada para reutilizar los embeddings.
vectores = modelo.encode(
    textos,
    convert_to_numpy=True,
    normalize_embeddings=True
)

vector = vectores[0]
print(f"Dimensiones del vector: {vector.shape[0]}")
print("Primeros 10 valores:", vector[:10])

graficar_embeddings(textos, vectores)
```
